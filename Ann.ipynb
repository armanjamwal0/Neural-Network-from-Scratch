```python
def perameter_init(layers):
    perameters = {}
    layer_dim = len(layers)
    for i in range(1,layer_dim):
        perameters[f'w{i}'] = np.ones((layers[i-1],layers[i]))*0.1
        perameters[f'b{i}'] = np.zeros((layers[i],1))
        # print(i)
    return perameters
```

```python
def linear(w,b,prv_layer):
    y_hat = np.dot(w.T,prv_layer) + b
    return y_hat
```

```python
def forward_propagation(X,perameters):
    total_layer = len(perameters) // 2
    A = X 
    prv_layer_output = {}
    counter = 0
    output = 1
    for i in range(1,total_layer+1):
        if counter == 0 :
            prv_layer = A
            w = perameters[f'w{i}']
            b = perameters[f'b{i}']
            
            A = linear(w,b,prv_layer)
            counter+=1
        else: 
            prv_layer_output[f'o{i}'] = A
            w = perameters[f'w{i}']
            b = perameters[f'b{i}']
            
            # print(prv_layer_output[f'o{i}'])
                      
            A = linear(w,b,prv_layer_output[f'o{i}'])  
            counter+=1
            output+=1
            
    return A,prv_layer_output
```

```python
def update_weights(perameter,prv_layer,y_hat,X,y,n):
    # Layer 1 -> part 1
    perameter['w1'][0][0] =  perameter['w1'][0][0] - n * (((-2 * (y -y_hat[0][0])) * perameter['w3'][0][0] * perameter['w2'][0][0] * X[0][0]) + ((-2 * (y -y_hat[0][0])) * perameter['w3'][1][0]*perameter['w2'][0][1]*X[0][0]))
    perameter['w1'][1][0] =  perameter['w1'][1][0] - n * (((-2 * (y -y_hat[0][0])) * perameter['w3'][0][0] * perameter['w2'][0][0] * X[1][0]) + ((-2 * (y -y_hat[0][0])) * perameter['w3'][1][0]*perameter['w2'][0][1]*X[1][0]))
    perameter['b1'][0][0] =  perameter['b1'][0][0] - n * ((-2 * (y -y_hat[0][0])) *  perameter['w3'][0][0] * perameter['w2'][0][0])           + ((-2 * (y -y_hat[0][0])) * perameter['w3'][1][0]*perameter['w2'][0][1])
    
    # Layer 1 -> part 2
    perameter['w1'][0][1] =  perameter['w1'][0][1] - n * (((-2 * (y -y_hat[0][0])) * perameter['w3'][0][0] * perameter['w2'][1][0] * X[0][0]) + ((-2 * (y -y_hat[0][0])) * perameter['w3'][1][0]*perameter['w2'][1][1]*X[0][0]))
    perameter['w1'][1][1] =  perameter['w1'][1][1] - n * (((-2 * (y -y_hat[0][0])) * perameter['w3'][0][0] * perameter['w2'][1][0] * X[1][0]) + ((-2 * (y -y_hat[0][0])) * perameter['w3'][1][0]*perameter['w2'][1][1]*X[1][0]))
    perameter['b1'][1][0] =  perameter['b1'][1][0] - n * (((-2 * (y -y_hat[0][0])) * perameter['w3'][0][0] * perameter['w2'][1][0])           + ((-2 * (y -y_hat[0][0])) * perameter['w3'][1][0]*perameter['w2'][1][1]))
    
    # Layer 1 -> part 3
    
    perameter['w1'][0][2] = perameter['w1'][0][2] - n * (((-2 * (y -y_hat[0][0])) * perameter['w3'][0][0] * perameter['w2'][2][0] * X[0][0]) + ((-2 * (y -y_hat[0][0])) * perameter['w3'][1][0]*perameter['w2'][2][1]*X[1][0]))
    perameter['w1'][1][2] = perameter['w1'][1][2] - n * (((-2 * (y -y_hat[0][0])) * perameter['w3'][0][0] * perameter['w2'][2][0] * X[0][0]) + ((-2 * (y -y_hat[0][0])) * perameter['w3'][1][0]*perameter['w2'][2][1]*X[1][0]))
    perameter['b1'][2][0] = perameter['b1'][2][0] - n * (((-2 * (y -y_hat[0][0])) * perameter['w3'][0][0] * perameter['w2'][2][0])           + ((-2 * (y -y_hat[0][0])) * perameter['w3'][1][0]*perameter['w2'][2][1]))
    
    # layer 2 part 1
    perameter['w2'][0][0] = perameter['w2'][0][0] -  n * (-2 *(y-y_hat[0][0]) * perameter['w3'][0][0]*prv_layer['o2'][0][0])
    perameter['w2'][1][0] = perameter['w2'][1][0] -  n * (-2 *(y-y_hat[0][0]) * perameter['w3'][0][0]*prv_layer['o2'][1][0])
    perameter['w2'][2][0] = perameter['w2'][2][0] -  n * (-2 *(y-y_hat[0][0]) * perameter['w3'][0][0]*prv_layer['o2'][2][0])
    perameter['b2'][0][0] = perameter['b2'][0][0] -  n * (-2 *(y-y_hat[0][0]) * perameter['w3'][0][0])
    
    # layer 2 part 2
    perameter['w2'][0][1] = perameter['w2'][0][1] - n * (-2 * (y-y_hat[0][0]) * perameter['w3'][1][0]*prv_layer['o2'][0][0])
    perameter['w2'][1][1] = perameter['w2'][1][1] - n * (-2 * (y-y_hat[0][0]) * perameter['w3'][1][0]*prv_layer['o2'][1][0])
    perameter['w2'][2][1] = perameter['w2'][2][1] - n * (-2 * (y-y_hat[0][0]) * perameter['w3'][1][0]*prv_layer['o2'][2][0])
    perameter['b2'][1][0] = perameter['b2'][1][0] - n * (-2 * (y-y_hat[0][0]) * perameter['w3'][1][0])
    
    #layer 3 
    perameter['w3'][0][0] = perameter['w3'][0][0] - n * (-2 * (y- y_hat[0][0]) * prv_layer['o3'][0][0])
    perameter['w3'][1][0] = perameter['w3'][1][0] - n * (-2 * (y- y_hat[0][0]) * prv_layer['o3'][1][0])
    perameter['b3'][0][0] = perameter['b3'][0][0] - n * (-2 * (y- y_hat[0][0]))
    
```

```python
perameter = perameter_init([2,3,2,1])
# y_hat,prv_layer = forward_propagation(X,perameter)
epochs = 100
for _ in range(epochs):
    loss = []
    for i in range(df.shape[0]):
        X = df.iloc[:,:-1].values[i].reshape(2,1)
        y = df.iloc[:,-1].values[i]
        y_hat,prv_layer = forward_propagation(X,perameter)
        # print(y)
        # print(y_hat)
        # print(X)
        # print(prv_layer)
        update_weights(perameter,prv_layer,y_hat,X,y,0.001)
        loss.append((y-y_hat)**2)
    print(f'Loss epochs {_} {np.mean(loss)}')
perameter
    
```
